```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")
df = df.drop_duplicates()
df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])
df = df.drop(columns=["deck"])
df["age"] = df["age"].round(0).astype(int)

print(df.shape)
print(df.describe())
```

```python
df["age_group"] = pd.cut(df["age"], bins=[0, 12, 18, 35, 60, 100],
                         labels=["Child", "Teen", "Young adult", "Adult", "Senior"])

print("Overall survival rate (%):", round(df["survived"].mean() * 100, 1))
print()
print("Survival rate by sex (%):")
print(df.groupby("sex")["survived"].mean().mul(100).round(1))
print()
print("Survival rate by class (%):")
print(df.groupby("pclass")["survived"].mean().mul(100).round(1))
print()
print("Average fare by class:")
print(df.groupby("pclass")["fare"].mean().round(1))
print()
print("Survival rate by age group (%):")
print(df.groupby("age_group", observed=True)["survived"].mean().mul(100).round(1))
print()
print("Highest fare:", df["fare"].max())
print("Passengers with fare 0:", (df["fare"] == 0).sum())
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 9))

sns.barplot(data=df, x="sex", y="survived", ax=axes[0, 0])
axes[0, 0].set_title("Survival Rate by Sex")

sns.barplot(data=df, x="pclass", y="survived", ax=axes[0, 1])
axes[0, 1].set_title("Survival Rate by Passenger Class")

sns.histplot(df["age"], bins=20, ax=axes[1, 0])
axes[1, 0].set_title("Age Distribution")

sns.barplot(data=df, x="age_group", y="survived", ax=axes[1, 1])
axes[1, 1].set_title("Survival Rate by Age Group")

plt.tight_layout()
plt.show()
```
